```python
# Get the user inputs for the first and second numbers.
num1 = float(input("Enter the first number."))
num2 = float(input("Enter the second number."))
choice = input("Select an Operation (addition, subtraction, multiplication, division):")

# Create a function for additiion.
def add_numbers(num1, num2):
  return num1 + num2

sum = add_numbers(num1, num2)
print(f"The sum of the two numbers is {sum}.")

# Create a function for subtraction.
def subtract_numbers(num1, num2):
  return num1 - num2

difference = subtract_numbers(num1, num2)

print(f"The difference of the two numbers is {difference}.")

# Create a function for multiplication.
def multiply_numbers(num1, num2):
  return num1 * num2

product = multiply_numbers(num1, num2)

print(f"The product of the two numbers is {product}.")

# Create a function for division.
def divide_numbers(num1, num2):
  return num1 / num2

quotient = divide_numbers(num1, num2)

if choice == "addition":
  output = add_numbers(num1, num2)
  print("sum =", output)

elif choice == "subtraction":
  output = subtract_numbers(num1, num2)
  print("sum =", output)

elif choice == "multiplication":
  output = multiply_numbers(num1, num2)
  print = ("sum =", output)

elif choice == "division":
  output = divide_numbers(num1, num2)
  print("sum =", output)

else:
  print("The selected operation is invalid.")
```
